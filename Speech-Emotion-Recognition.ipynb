```python
# ============================================================
# DEEP LEARNING-BASED SPEECH EMOTION RECOGNITION
# USING LSTM NETWORKS
#
# DATASET  : RAVDESS Speech Audio
# CLASSES  : 8
# MODEL    : BiLSTM + Attention
# FEATURES : MFCC + Delta + Delta-Delta
# ============================================================


# ============================================================
# 1. INSTALL REQUIRED LIBRARIES
# ============================================================

!pip install -q librosa soundfile scikit-learn seaborn kagglehub


# ============================================================
# 2. IMPORT LIBRARIES
# ============================================================

import os
import glob
import random
import pickle
import warnings
import shutil

import numpy as np
import pandas as pd

import librosa
import soundfile as sf

import matplotlib.pyplot as plt
import seaborn as sns

import tensorflow as tf

from tensorflow.keras.models import Model, load_model
from tensorflow.keras.layers import (
    Input,
    LSTM,
    Bidirectional,
    Dense,
    Dropout,
    LayerNormalization,
    BatchNormalization,
    Attention,
    GlobalAveragePooling1D
)

from tensorflow.keras.callbacks import (
    EarlyStopping,
    ReduceLROnPlateau,
    ModelCheckpoint
)

from tensorflow.keras.optimizers import Adam

from sklearn.preprocessing import StandardScaler
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

from sklearn.utils.class_weight import compute_class_weight

warnings.filterwarnings("ignore")


# ============================================================
# 3. REPRODUCIBILITY
# ============================================================

SEED = 42

random.seed(SEED)
np.random.seed(SEED)
tf.random.set_seed(SEED)

os.environ["PYTHONHASHSEED"] = str(SEED)

print("Random seed:", SEED)


# ============================================================
# 4. CHECK TENSORFLOW AND GPU
# ============================================================

print("=" * 70)
print("SYSTEM INFORMATION")
print("=" * 70)

print("TensorFlow version:", tf.__version__)

gpus = tf.config.list_physical_devices("GPU")

print("Available GPUs:", gpus)

if gpus:

    print("\n✅ GPU DETECTED")

    for gpu in gpus:
        print(gpu)

else:

    print("\n⚠️ NO GPU DETECTED")
    print("Go to:")
    print("Runtime → Change runtime type → GPU")


# ============================================================
# 5. SHOW NVIDIA GPU INFORMATION
# ============================================================

print("\n" + "=" * 70)
print("NVIDIA GPU INFORMATION")
print("=" * 70)

!nvidia-smi


# ============================================================
# 6. DOWNLOAD RAVDESS FROM KAGGLE
# ============================================================

import kagglehub

print("\n" + "=" * 70)
print("DOWNLOADING RAVDESS DATASET")
print("=" * 70)

path = kagglehub.dataset_download(
    "uwrfkaggler/ravdess-emotional-speech-audio"
)

print("Kaggle dataset path:")
print(path)


# ============================================================
# 7. FIND audio_speech_actors_01-24
# ============================================================

print("\n" + "=" * 70)
print("SEARCHING FOR DATASET FOLDER")
print("=" * 70)

target_folder = None

for root, dirs, files in os.walk(path):

    if "audio_speech_actors_01-24" in dirs:

        target_folder = os.path.join(
            root,
            "audio_speech_actors_01-24"
        )

        break


# ------------------------------------------------------------
# FALLBACK
# ------------------------------------------------------------

if target_folder is None:

    actor_folders = [
        folder
        for folder in os.listdir(path)
        if folder.startswith("Actor_")
    ]

    if len(actor_folders) >= 20:

        target_folder = path


if target_folder is None:

    print("\nDataset structure found:")

    for root, dirs, files in os.walk(path):

        print(root)

        if len(files) > 0:

            print("Files:", files[:5])


    raise FileNotFoundError(
        "audio_speech_actors_01-24 folder was not found."
    )


dataset_path = target_folder

print("\n✅ Dataset folder found:")
print(dataset_path)


# ============================================================
# 8. FIND ACTOR FOLDERS
# ============================================================

actors = sorted(
    [
        folder
        for folder in os.listdir(dataset_path)
        if folder.startswith("Actor_")
    ]
)

print("\n" + "=" * 70)
print("ACTOR INFORMATION")
print("=" * 70)

print("Number of actors:", len(actors))
print("Actors:")

for actor in actors:
    print(actor)


# ============================================================
# 9. COUNT WAV FILES
# ============================================================

wav_files = glob.glob(
    os.path.join(
        dataset_path,
        "**",
        "*.wav"
    ),
    recursive=True
)

print("\n" + "=" * 70)
print("DATASET INFORMATION")
print("=" * 70)

print("Total WAV files:", len(wav_files))


if len(actors) == 24:
    print("✅ 24 actors confirmed")
else:
    print("⚠️ Actor count:", len(actors))


if len(wav_files) == 1440:
    print("✅ 1440 WAV files confirmed")
else:
    print("⚠️ WAV count:", len(wav_files))


# ============================================================
# 10. EMOTION MAPPING
# ============================================================

emotion_map = {

    1: "neutral",
    2: "calm",
    3: "happy",
    4: "sad",
    5: "angry",
    6: "fearful",
    7: "disgust",
    8: "surprised"

}

class_names = [

    "neutral",
    "calm",
    "happy",
    "sad",
    "angry",
    "fearful",
    "disgust",
    "surprised"

]

NUM_CLASSES = len(class_names)

print("\n" + "=" * 70)
print("EMOTION CLASSES")
print("=" * 70)

for key, value in emotion_map.items():

    print(
        key,
        "→",
        value
    )


# ============================================================
# 11. CREATE METADATA
# ============================================================

print("\n" + "=" * 70)
print("CREATING METADATA")
print("=" * 70)

data = []

for actor_folder in actors:

    actor_id = int(
        actor_folder.split("_")[1]
    )

    actor_path = os.path.join(
        dataset_path,
        actor_folder
    )

    files = os.listdir(actor_path)

    for file in files:

        if file.lower().endswith(".wav"):

            filename = file.replace(
                ".wav",
                ""
            )

            parts = filename.split("-")

            if len(parts) != 7:

                continue

            emotion_id = int(parts[2])

            if emotion_id not in emotion_map:

                continue

            emotion = emotion_map[
                emotion_id
            ]

            data.append({

                "file_path":
                    os.path.join(
                        actor_path,
                        file
                    ),

                "emotion_id":
                    emotion_id,

                "emotion":
                    emotion,

                "actor":
                    actor_id

            })


df = pd.DataFrame(data)

print("Total metadata samples:", len(df))

display(df.head())


# ============================================================
# 12. CHECK MISSING VALUES
# ============================================================

print("\nMissing values:")

print(
    df.isnull().sum()
)


# ============================================================
# 13. EMOTION DISTRIBUTION
# ============================================================

print("\n" + "=" * 70)
print("EMOTION DISTRIBUTION")
print("=" * 70)

emotion_counts = (
    df["emotion"]
    .value_counts()
    .reindex(class_names)
)

print(emotion_counts)


# ============================================================
# 14. PLOT EMOTION DISTRIBUTION
# ============================================================

plt.figure(
    figsize=(10, 6)
)

emotion_counts.plot(
    kind="bar"
)

plt.title(
    "RAVDESS Emotion Distribution"
)

plt.xlabel(
    "Emotion"
)

plt.ylabel(
    "Number of Audio Files"
)

plt.xticks(
    rotation=45
)

plt.tight_layout()

plt.show()


# ============================================================
# 15. SPEAKER-INDEPENDENT SPLIT
# ============================================================

print("\n" + "=" * 70)
print("SPEAKER-INDEPENDENT SPLIT")
print("=" * 70)

# Actors 01-16 → TRAIN
# Actors 17-20 → VALIDATION
# Actors 21-24 → TEST

train_df = df[
    df["actor"].between(
        1,
        16
    )
].copy()

val_df = df[
    df["actor"].between(
        17,
        20
    )
].copy()

test_df = df[
    df["actor"].between(
        21,
        24
    )
].copy()


print(
    "Training samples:",
    len(train_df)
)

print(
    "Validation samples:",
    len(val_df)
)

print(
    "Testing samples:",
    len(test_df)
)


print("\nTraining actors:")
print(
    sorted(
        train_df["actor"].unique()
    )
)

print("\nValidation actors:")
print(
    sorted(
        val_df["actor"].unique()
    )
)

print("\nTesting actors:")
print(
    sorted(
        test_df["actor"].unique()
    )
)


# ============================================================
# 16. AUDIO CONFIGURATION
# ============================================================

SAMPLE_RATE = 16000

N_MFCC = 40

MAX_LEN = 250

HOP_LENGTH = 256

N_FFT = 512

print("\n" + "=" * 70)
print("AUDIO CONFIGURATION")
print("=" * 70)

print("Sample rate:", SAMPLE_RATE)

print("MFCC:", N_MFCC)

print("Maximum sequence length:", MAX_LEN)

print("Hop length:", HOP_LENGTH)

print("FFT:", N_FFT)

print("Total features per frame:", N_MFCC * 3)


# ============================================================
# 17. FEATURE EXTRACTION FUNCTION
# ============================================================

def extract_features(
    file_path,
    augment=False
):

    try:

        # ----------------------------------------------------
        # LOAD AUDIO
        # ----------------------------------------------------

        audio, sr = librosa.load(
            file_path,
            sr=SAMPLE_RATE,
            mono=True
        )

        # ----------------------------------------------------
        # TRIM SILENCE
        # ----------------------------------------------------

        audio, _ = librosa.effects.trim(
            audio,
            top_db=25
        )

        # ----------------------------------------------------
        # HANDLE EMPTY AUDIO
        # ----------------------------------------------------

        if len(audio) == 0:

            audio = np.zeros(
                SAMPLE_RATE,
                dtype=np.float32
            )

        # ----------------------------------------------------
        # NORMALIZATION
        # ----------------------------------------------------

        max_amp = np.max(
            np.abs(audio)
        )

        if max_amp > 0:

            audio = (
                audio / max_amp
            )

        # ----------------------------------------------------
        # DATA AUGMENTATION
        # ONLY FOR TRAINING
        # ----------------------------------------------------

        if augment:

            # Random noise
            noise_level = np.random.uniform(
                0.001,
                0.006
            )

            noise = np.random.normal(
                0,
                1,
                len(audio)
            )

            audio = (
                audio +
                noise_level * noise
            )

            # Random time shift
            shift = np.random.randint(
                -1600,
                1600
            )

            audio = np.roll(
                audio,
                shift
            )

            # Random volume
            volume = np.random.uniform(
                0.85,
                1.15
            )

            audio = (
                audio * volume
            )

            # Clip
            audio = np.clip(
                audio,
                -1,
                1
            )

        # ----------------------------------------------------
        # MFCC
        # ----------------------------------------------------

        mfcc = librosa.feature.mfcc(
            y=audio,
            sr=sr,
            n_mfcc=N_MFCC,
            n_fft=N_FFT,
            hop_length=HOP_LENGTH
        )

        # ----------------------------------------------------
        # DELTA
        # ----------------------------------------------------

        delta = librosa.feature.delta(
            mfcc
        )

        # ----------------------------------------------------
        # DELTA-DELTA
        # ----------------------------------------------------

        delta_delta = librosa.feature.delta(
            mfcc,
            order=2
        )

        # ----------------------------------------------------
        # COMBINE
        # ----------------------------------------------------

        features = np.concatenate(
            [
                mfcc,
                delta,
                delta_delta
            ],
            axis=0
        )

        # ----------------------------------------------------
        # FIXED LENGTH
        # ----------------------------------------------------

        current_length = features.shape[1]

        if current_length < MAX_LEN:

            pad_width = (
                MAX_LEN -
                current_length
            )

            features = np.pad(
                features,
                (
                    (0, 0),
                    (0, pad_width)
                ),
                mode="constant"
            )

        else:

            features = (
                features[
                    :,
                    :MAX_LEN
                ]
            )

        # ----------------------------------------------------
        # TRANSPOSE
        # ----------------------------------------------------

        features = features.T

        return features.astype(
            np.float32
        )

    except Exception as e:

        print(
            "Feature extraction error:",
            file_path,
            e
        )

        return None


# ============================================================
# 18. TEST FEATURE EXTRACTION
# ============================================================

print("\n" + "=" * 70)
print("TESTING FEATURE EXTRACTION")
print("=" * 70)

sample_features = extract_features(
    df.iloc[0]["file_path"]
)

print(
    "Feature shape:",
    sample_features.shape
)

print(
    "Expected shape:",
    (
        MAX_LEN,
        N_MFCC * 3
    )
)


# ============================================================
# 19. PREPARE DATASET FUNCTION
# ============================================================

def prepare_dataset(
    dataframe,
    augment=False
):

    X = []
    y = []

    total = len(dataframe)

    for i, (_, row) in enumerate(
        dataframe.iterrows()
    ):

        if i % 100 == 0:

            print(
                f"Processing "
                f"{i}/{total}"
            )

        features = extract_features(
            row["file_path"],
            augment=augment
        )

        if features is not None:

            X.append(
                features
            )

            y.append(
                row["emotion_id"] - 1
            )

    X = np.array(
        X,
        dtype=np.float32
    )

    y = np.array(
        y,
        dtype=np.int32
    )

    return X, y


# ============================================================
# 20. EXTRACT ORIGINAL TRAINING FEATURES
# ============================================================

print("\n" + "=" * 70)
print("EXTRACTING TRAINING FEATURES")
print("=" * 70)

X_train_original, y_train_original = (
    prepare_dataset(
        train_df,
        augment=False
    )
)

print(
    "Original training X:",
    X_train_original.shape
)

print(
    "Original training y:",
    y_train_original.shape
)


# ============================================================
# 21. EXTRACT AUGMENTED TRAINING FEATURES
# ============================================================

print("\n" + "=" * 70)
print("EXTRACTING AUGMENTED TRAINING FEATURES")
print("=" * 70)

X_train_aug, y_train_aug = (
    prepare_dataset(
        train_df,
        augment=True
    )
)

print(
    "Augmented training X:",
    X_train_aug.shape
)

print(
    "Augmented training y:",
    y_train_aug.shape
)


# ============================================================
# 22. COMBINE ORIGINAL + AUGMENTED
# ============================================================

X_train = np.concatenate(
    [
        X_train_original,
        X_train_aug
    ],
    axis=0
)

y_train = np.concatenate(
    [
        y_train_original,
        y_train_aug
    ],
    axis=0
)

print("\n" + "=" * 70)
print("FINAL TRAINING DATA")
print("=" * 70)

print(
    "X_train:",
    X_train.shape
)

print(
    "y_train:",
    y_train.shape
)


# ============================================================
# 23. VALIDATION FEATURES
# ============================================================

print("\n" + "=" * 70)
print("EXTRACTING VALIDATION FEATURES")
print("=" * 70)

X_val, y_val = prepare_dataset(
    val_df,
    augment=False
)

print(
    "X_val:",
    X_val.shape
)

print(
    "y_val:",
    y_val.shape
)


# ============================================================
# 24. TEST FEATURES
# ============================================================

print("\n" + "=" * 70)
print("EXTRACTING TEST FEATURES")
print("=" * 70)

X_test, y_test = prepare_dataset(
    test_df,
    augment=False
)

print(
    "X_test:",
    X_test.shape
)

print(
    "y_test:",
    y_test.shape
)


# ============================================================
# 25. FEATURE STANDARDIZATION
# ============================================================

print("\n" + "=" * 70)
print("FEATURE STANDARDIZATION")
print("=" * 70)

num_features = X_train.shape[2]

scaler = StandardScaler()

# ------------------------------------------------------------
# FIT ONLY ON TRAINING DATA
# ------------------------------------------------------------

X_train_2d = X_train.reshape(
    -1,
    num_features
)

scaler.fit(
    X_train_2d
)

# ------------------------------------------------------------
# TRANSFORM TRAINING
# ------------------------------------------------------------

X_train = scaler.transform(
    X_train_2d
).reshape(
    X_train.shape
)

# ------------------------------------------------------------
# TRANSFORM VALIDATION
# ------------------------------------------------------------

X_val = scaler.transform(
    X_val.reshape(
        -1,
        num_features
    )
).reshape(
    X_val.shape
)

# ------------------------------------------------------------
# TRANSFORM TEST
# ------------------------------------------------------------

X_test = scaler.transform(
    X_test.reshape(
        -1,
        num_features
    )
).reshape(
    X_test.shape
)

print("Scaling completed.")

print(
    "X_train:",
    X_train.shape
)

print(
    "X_val:",
    X_val.shape
)

print(
    "X_test:",
    X_test.shape
)


# ============================================================
# 26. SHUFFLE TRAINING DATA
# ============================================================

indices = np.random.permutation(
    len(X_train)
)

X_train = X_train[
    indices
]

y_train = y_train[
    indices
]

print(
    "Training data shuffled."
)


# ============================================================
# 27. CLASS WEIGHTS
# ============================================================

print("\n" + "=" * 70)
print("CALCULATING CLASS WEIGHTS")
print("=" * 70)

classes = np.unique(
    y_train
)

weights = compute_class_weight(
    class_weight="balanced",
    classes=classes,
    y=y_train
)

class_weights = {
    int(cls): float(weight)
    for cls, weight in zip(
        classes,
        weights
    )
}

print(
    class_weights
)


# ============================================================
# 28. CLEAR PREVIOUS MODEL
# ============================================================

tf.keras.backend.clear_session()


# ============================================================
# 29. BUILD IMPROVED BiLSTM + ATTENTION MODEL
# ============================================================

print("\n" + "=" * 70)
print("BUILDING MODEL")
print("=" * 70)

input_shape = (
    MAX_LEN,
    N_MFCC * 3
)

inputs = Input(
    shape=input_shape,
    name="MFCC_Input"
)


# ------------------------------------------------------------
# FIRST BiLSTM
# ------------------------------------------------------------

x = Bidirectional(
    LSTM(
        128,
        return_sequences=True,
        dropout=0.2,
        recurrent_dropout=0.0
    ),
    name="BiLSTM_128"
)(inputs)


# ------------------------------------------------------------
# NORMALIZATION
# ------------------------------------------------------------

x = LayerNormalization(
    name="LayerNorm_1"
)(x)


# ------------------------------------------------------------
# SECOND BiLSTM
# ------------------------------------------------------------

x = Bidirectional(
    LSTM(
        64,
        return_sequences=True,
        dropout=0.2,
        recurrent_dropout=0.0
    ),
    name="BiLSTM_64"
)(x)


# ------------------------------------------------------------
# NORMALIZATION
# ------------------------------------------------------------

x = LayerNormalization(
    name="LayerNorm_2"
)(x)


# ------------------------------------------------------------
# ATTENTION
# ------------------------------------------------------------

attention_output = Attention(
    name="Self_Attention"
)([
    x,
    x
])


# ------------------------------------------------------------
# GLOBAL AVERAGE POOLING
# ------------------------------------------------------------

x = GlobalAveragePooling1D(
    name="Global_Average_Pooling"
)(
    attention_output
)


# ------------------------------------------------------------
# DENSE LAYER
# ------------------------------------------------------------

x = Dense(
    128,
    activation="relu",
    name="Dense_128"
)(x)


# ------------------------------------------------------------
# BATCH NORMALIZATION
# ------------------------------------------------------------

x = BatchNormalization(
    name="BatchNorm"
)(x)


# ------------------------------------------------------------
# DROPOUT
# ------------------------------------------------------------

x = Dropout(
    0.4,
    name="Dropout"
)(x)


# ------------------------------------------------------------
# OUTPUT
# ------------------------------------------------------------

outputs = Dense(
    NUM_CLASSES,
    activation="softmax",
    name="Emotion_Output"
)(x)


# ------------------------------------------------------------
# CREATE MODEL
# ------------------------------------------------------------

model = Model(
    inputs=inputs,
    outputs=outputs,
    name="Speech_Emotion_BiLSTM_Attention"
)


# ============================================================
# 30. MODEL SUMMARY
# ============================================================

model.summary()


# ============================================================
# 31. COMPILE MODEL
# ============================================================

optimizer = Adam(
    learning_rate=0.0005
)

model.compile(
    optimizer=optimizer,
    loss="sparse_categorical_crossentropy",
    metrics=[
        "accuracy"
    ]
)

print(
    "\n✅ Model compiled successfully."
)


# ============================================================
# 32. CALLBACKS
# ============================================================

checkpoint_path = (
    "/content/"
    "best_speech_emotion_lstm.keras"
)

checkpoint = ModelCheckpoint(
    checkpoint_path,
    monitor="val_accuracy",
    mode="max",
    save_best_only=True,
    verbose=1
)


reduce_lr = ReduceLROnPlateau(
    monitor="val_loss",
    factor=0.5,
    patience=4,
    min_lr=1e-6,
    verbose=1
)


early_stopping = EarlyStopping(
    monitor="val_loss",
    patience=10,
    restore_best_weights=True,
    verbose=1
)


# ============================================================
# 33. TRAIN MODEL
# ============================================================

print("\n" + "=" * 70)
print("STARTING TRAINING")
print("=" * 70)

history = model.fit(

    X_train,
    y_train,

    validation_data=(
        X_val,
        y_val
    ),

    epochs=50,

    batch_size=32,

    class_weight=class_weights,

    callbacks=[
        checkpoint,
        reduce_lr,
        early_stopping
    ],

    verbose=1
)

print("\n" + "=" * 70)
print("TRAINING FINISHED")
print("=" * 70)


# ============================================================
# 34. NUMBER OF ACTUAL EPOCHS
# ============================================================

actual_epochs = len(
    history.history["loss"]
)

print(
    "Actual epochs completed:",
    actual_epochs
)

print(
    "Maximum epochs:",
    50
)


# ============================================================
# 35. LOAD BEST MODEL
# ============================================================

print("\n" + "=" * 70)
print("LOADING BEST MODEL")
print("=" * 70)

model = load_model(
    checkpoint_path
)

print(
    "✅ Best model loaded."
)


# ============================================================
# 36. TRAINING ACCURACY GRAPH
# ============================================================

plt.figure(
    figsize=(10, 6)
)

plt.plot(
    history.history[
        "accuracy"
    ],
    label="Training Accuracy"
)

plt.plot(
    history.history[
        "val_accuracy"
    ],
    label="Validation Accuracy"
)

plt.title(
    "Training and Validation Accuracy"
)

plt.xlabel(
    "Epoch"
)

plt.ylabel(
    "Accuracy"
)

plt.legend()

plt.grid(
    alpha=0.3
)

plt.tight_layout()

plt.show()


# ============================================================
# 37. TRAINING LOSS GRAPH
# ============================================================

plt.figure(
    figsize=(10, 6)
)

plt.plot(
    history.history[
        "loss"
    ],
    label="Training Loss"
)

plt.plot(
    history.history[
        "val_loss"
    ],
    label="Validation Loss"
)

plt.title(
    "Training and Validation Loss"
)

plt.xlabel(
    "Epoch"
)

plt.ylabel(
    "Loss"
)

plt.legend()

plt.grid(
    alpha=0.3
)

plt.tight_layout()

plt.show()


# ============================================================
# 38. FINAL TEST EVALUATION
# ============================================================

print("\n" + "=" * 70)
print("FINAL TEST EVALUATION")
print("=" * 70)

test_loss, test_accuracy = model.evaluate(
    X_test,
    y_test,
    verbose=1
)

print(
    "\nTest Loss:",
    round(
        test_loss,
        4
    )
)

print(
    "Test Accuracy:",
    round(
        test_accuracy * 100,
        2
    ),
    "%"
)


# ============================================================
# 39. GENERATE TEST PREDICTIONS
# ============================================================

print("\nGenerating predictions...")

y_pred_prob = model.predict(
    X_test,
    verbose=1
)

y_pred = np.argmax(
    y_pred_prob,
    axis=1
)


# ============================================================
# 40. OVERALL METRICS
# ============================================================

accuracy = accuracy_score(
    y_test,
    y_pred
)

precision = precision_score(
    y_test,
    y_pred,
    average="weighted",
    zero_division=0
)

recall = recall_score(
    y_test,
    y_pred,
    average="weighted",
    zero_division=0
)

f1 = f1_score(
    y_test,
    y_pred,
    average="weighted",
    zero_division=0
)


print("\n" + "=" * 70)
print("FINAL METRICS")
print("=" * 70)

print(
    f"Accuracy  : {accuracy * 100:.2f}%"
)

print(
    f"Precision : {precision * 100:.2f}%"
)

print(
    f"Recall    : {recall * 100:.2f}%"
)

print(
    f"F1 Score  : {f1 * 100:.2f}%"
)


# ============================================================
# 41. CLASSIFICATION REPORT
# ============================================================

print("\n" + "=" * 70)
print("CLASSIFICATION REPORT")
print("=" * 70)

report = classification_report(
    y_test,
    y_pred,
    target_names=class_names,
    digits=4,
    zero_division=0
)

print(report)


# ============================================================
# 42. CONFUSION MATRIX
# ============================================================

cm = confusion_matrix(
    y_test,
    y_pred
)

plt.figure(
    figsize=(10, 8)
)

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    xticklabels=class_names,
    yticklabels=class_names
)

plt.title(
    "Confusion Matrix - BiLSTM Attention"
)

plt.xlabel(
    "Predicted Emotion"
)

plt.ylabel(
    "Actual Emotion"
)

plt.tight_layout()

plt.show()


# ============================================================
# 43. PER-CLASS ACCURACY
# ============================================================

print("\n" + "=" * 70)
print("PER-CLASS ACCURACY")
print("=" * 70)

for i, emotion in enumerate(
    class_names
):

    total = np.sum(
        y_test == i
    )

    correct = np.sum(
        (y_test == i) &
        (y_pred == i)
    )

    if total > 0:

        class_accuracy = (
            correct / total
        ) * 100

    else:

        class_accuracy = 0

    print(
        f"{emotion:12s}: "
        f"{class_accuracy:.2f}%"
    )


# ============================================================
# 44. SAVE FINAL MODEL
# ============================================================

final_model_path = (
    "/content/"
    "speech_emotion_lstm.keras"
)

model.save(
    final_model_path
)

print(
    "\n✅ Model saved:"
)

print(
    final_model_path
)


# ============================================================
# 45. SAVE FEATURE SCALER
# ============================================================

scaler_path = (
    "/content/"
    "feature_scaler.pkl"
)

with open(
    scaler_path,
    "wb"
) as f:

    pickle.dump(
        scaler,
        f
    )

print(
    "✅ Scaler saved:"
)

print(
    scaler_path
)


# ============================================================
# 46. SAVE EMOTION LABELS
# ============================================================

labels_path = (
    "/content/"
    "emotion_labels.pkl"
)

with open(
    labels_path,
    "wb"
) as f:

    pickle.dump(
        class_names,
        f
    )

print(
    "✅ Labels saved:"
)

print(
    labels_path
)


# ============================================================
# 47. SAVE METADATA
# ============================================================

metadata_path = (
    "/content/"
    "ravdess_metadata.csv"
)

df.to_csv(
    metadata_path,
    index=False
)

print(
    "✅ Metadata saved:"
)

print(
    metadata_path
)


# ============================================================
# 48. SAVE TRAINING HISTORY
# ============================================================

history_path = (
    "/content/"
    "training_history.pkl"
)

with open(
    history_path,
    "wb"
) as f:

    pickle.dump(
        history.history,
        f
    )

print(
    "✅ Training history saved."
)


# ============================================================
# 49. SAVE CONFUSION MATRIX
# ============================================================

plt.figure(
    figsize=(10, 8)
)

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    xticklabels=class_names,
    yticklabels=class_names
)

plt.title(
    "Confusion Matrix - Speech Emotion Recognition"
)

plt.xlabel(
    "Predicted Emotion"
)

plt.ylabel(
    "Actual Emotion"
)

plt.tight_layout()

confusion_matrix_path = (
    "/content/"
    "confusion_matrix.png"
)

plt.savefig(
    confusion_matrix_path,
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print(
    "Confusion matrix saved:"
)

print(
    confusion_matrix_path
)


# ============================================================
# 50. SAVE ACCURACY GRAPH
# ============================================================

plt.figure(
    figsize=(10, 6)
)

plt.plot(
    history.history[
        "accuracy"
    ],
    label="Training Accuracy"
)

plt.plot(
    history.history[
        "val_accuracy"
    ],
    label="Validation Accuracy"
)

plt.title(
    "Training and Validation Accuracy"
)

plt.xlabel(
    "Epoch"
)

plt.ylabel(
    "Accuracy"
)

plt.legend()

plt.grid(
    alpha=0.3
)

plt.tight_layout()

accuracy_graph_path = (
    "/content/"
    "accuracy_graph.png"
)

plt.savefig(
    accuracy_graph_path,
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# 51. SAVE LOSS GRAPH
# ============================================================

plt.figure(
    figsize=(10, 6)
)

plt.plot(
    history.history[
        "loss"
    ],
    label="Training Loss"
)

plt.plot(
    history.history[
        "val_loss"
    ],
    label="Validation Loss"
)

plt.title(
    "Training and Validation Loss"
)

plt.xlabel(
    "Epoch"
)

plt.ylabel(
    "Loss"
)

plt.legend()

plt.grid(
    alpha=0.3
)

plt.tight_layout()

loss_graph_path = (
    "/content/"
    "loss_graph.png"
)

plt.savefig(
    loss_graph_path,
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# 52. FINAL SUMMARY
# ============================================================

print("\n")
print("=" * 70)
print("PROJECT TRAINING SUMMARY")
print("=" * 70)

print(
    "Dataset          : RAVDESS"
)

print(
    "Total files      :",
    len(df)
)

print(
    "Training files   :",
    len(train_df)
)

print(
    "Validation files :",
    len(val_df)
)

print(
    "Testing files    :",
    len(test_df)
)

print(
    "Classes          :",
    NUM_CLASSES
)

print(
    "Features         : MFCC + Delta + Delta-Delta"
)

print(
    "Model            : BiLSTM + Attention"
)

print(
    "Maximum epochs   : 50"
)

print(
    "Actual epochs    :",
    actual_epochs
)

print(
    f"Test Accuracy    : {accuracy * 100:.2f}%"
)

print(
    f"Precision        : {precision * 100:.2f}%"
)

print(
    f"Recall           : {recall * 100:.2f}%"
)

print(
    f"F1 Score         : {f1 * 100:.2f}%"
)

print("=" * 70)


# ============================================================
# 53. DOWNLOAD TRAINED MODEL
# ============================================================

from google.colab import files

print(
    "\nDownloading trained model..."
)

files.download(
    final_model_path
)


# ============================================================
# 54. DOWNLOAD SCALER
# ============================================================

files.download(
    scaler_path
)


# ============================================================
# 55. DOWNLOAD LABELS
# ============================================================

files.download(
    labels_path
)

print(
    "\n✅ TRAINING PIPELINE COMPLETED!"
)
```
